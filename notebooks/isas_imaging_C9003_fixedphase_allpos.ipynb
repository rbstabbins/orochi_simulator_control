```python
sns.set_context("paper", font_scale=1.1)
sns.set_style('ticks')
```

```python
def format_plot(fig, ax, results, show_spatial_stddev, scene):
    notation = 'R'

    cam_cols = channel_cols()

    min_550 = results[results['cwl'] == 550]['Mean of Single-Frame ROI'].idxmin()
    for i in np.arange(0, len(results)):
        lo_cwl = results['cwl'][i] - results['fwhm'][i]/2
        hi_cwl = results['cwl'][i] + results['fwhm'][i]/2
        ax.axvspan(lo_cwl, hi_cwl, alpha=0.2, color=cam_cols[results['Device'][i]])    
        # get ylim
        ymin, ymax = ax.get_ylim()
        x_place = results['cwl'][i] - results['fwhm'][i]/2
        y_place = ymax
        if i == min_550:
            x_place = results['cwl'][i] + results['fwhm'][i]/2
            y_place = ymin                

        # if show_spatial_stddev:
        #     error = results['Std. Dev. of Single-Frame ROI'][i]
        # else:
        #     error = results['Mean of Single-Frame Noise ROI'][i]        
        # annotate with phase angle
        ax.annotate(f'$g$: {results["Phase"][i]:.1f}°', 
                    (x_place, y_place), 
                    xytext=(0.05,0.05), 
                    textcoords='offset fontsize', 
                    fontsize='x-small', 
                    ha='left', va='bottom') 
        # annotate with emission angle
        ax.annotate(f'$e$: {results["Emission"][i]:.1f}°', 
                    (x_place, y_place), 
                    xytext=(0.05,1.05), 
                    textcoords='offset fontsize', 
                    fontsize='x-small', 
                    ha='left', va='bottom')
        # annotate with azimuth angle
        ax.annotate(rf'$\psi$: {results["Azimuth"][i]:.1f}°', 
                    (x_place, y_place), 
                    xytext=(0.05,2.05), 
                    textcoords='offset fontsize', 
                    fontsize='x-small', 
                    ha='left', va='bottom')

    # ax.legend(fontsize='x-small', bbox_to_anchor=(1.01, 0.5), loc='center left')
    ax.legend(fontsize='x-small', loc='lower right')

    ax.set_xlabel('Wavelength (nm)')
    ax.set_ylabel('Reflectance Factor')

    sns.despine(ax=ax)

    ax.tick_params(reset=True)
    ax.tick_params(which='both', length=5, direction='out', top=False, right=False, left=True, bottom=True)

    fig.tight_layout()

    return fig, ax
```

```python
def compare_plot(fig, ax, results, show_spatial_stddev, scene):

    notation = 'R'
    
    # Mean Reflectance ± Mean of Noise
    ax.errorbar(    
        x = results['cwl'],
        y = results['Mean of Single-Frame ROI'],
        yerr = results['Mean of Single-Frame Noise ROI'],
        label = rf'$\mu_{notation} \pm \mu_{{\sigma_{notation}}}$ {scene}',
        fmt='o-',    
        # ecolor='k',
        capsize=5
    )

    # Mean reflectance ± Standard Error of Reflectance - perhaps remove this?
    if show_spatial_stddev:
        # Mean reflectance ± Standard Deviation of Reflectance
        ax.errorbar(    
            x = results['cwl'],
            y = results['Mean of Single-Frame ROI'],
            yerr = results['Std. Dev. of Single-Frame ROI'],
            # label = fr'$\mu_{notation}\pm s_{notation}$ {scene}',  
            marker='none',
            linestyle='none', 
            ecolor='k',
            capsize=2
        )    
    
    return fig, ax
```

```python
fig, ax = plt.subplots(figsize=(5,3.4), dpi=300)

fig, ax = compare_plot(fig, ax, pos1_results, show_spatial_stddev, 'C9003 (1)')
fig, ax = compare_plot(fig, ax, pos2_results, show_spatial_stddev, 'C9003 (2)')
fig, ax = compare_plot(fig, ax, pos3_results, show_spatial_stddev, 'C9003 (3)')
fig, ax = format_plot(fig, ax, pos1_results, show_spatial_stddev, 'C9003 (1)')

fig.savefig(Path(f'C9003_pos123_reflectance.pdf'))
```

```python
def compare_plot_normalised(fig, ax, results, show_spatial_stddev, scene):

    notation = 'R'
    
    # Mean Reflectance ± Mean of Noise
    ax.errorbar(    
        x = results['cwl'],
        y = results['Mean of Single-Frame ROI Normalised'],
        yerr = results['Mean of Single-Frame Noise ROI Normalised'],
        label = rf'$\mu_{notation} \pm \mu_{{\sigma_{notation}}}$ {scene}',
        fmt='o-',    
        # ecolor='k',
        capsize=5
    )

    # Mean reflectance ± Standard Error of Reflectance - perhaps remove this?
    if show_spatial_stddev:
        # Mean reflectance ± Standard Deviation of Reflectance
        ax.errorbar(    
            x = results['cwl'],
            y = results['Mean of Single-Frame ROI Normalised'],
            yerr = results['Std. Dev. of Single-Frame ROI Normalised'],
            # label = fr'$\mu_{notation}\pm s_{notation}$ {scene}',  
            marker='none',
            linestyle='none', 
            ecolor='k',
            capsize=2
        )    
    
    return fig, ax
```

```python
fig, ax = plt.subplots(figsize=(5,3.4), dpi=300)

fig, ax = compare_plot_normalised(fig, ax, pos1_results, show_spatial_stddev, 'C9003 (1)')
fig, ax = compare_plot_normalised(fig, ax, pos2_results, show_spatial_stddev, 'C9003 (2)')
fig, ax = compare_plot_normalised(fig, ax, pos3_results, show_spatial_stddev, 'C9003 (3)')
fig, ax = format_plot(fig, ax, pos1_results, show_spatial_stddev, 'C9003 (1)')
ax.set_ylabel('Normalised Reflectance Factor')

fig.savefig(Path(f'C9003_pos123_reflectance_normalised.pdf'))
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np

def plot_reflectance_vs_cwl_per_step(concat_results, normalise):

    fig, ax = plt.subplots(figsize=(5,3.4), dpi=300)
    # add grid
    # plt.grid(visible=True)

    color = plt.cm.twilight_shifted(np.linspace(0, 1, len(steps)))

    for c, step in enumerate(steps):
        step_results = concat_results[concat_results['step'] == step].copy()
        step_results.sort_values(by='cwl', inplace=True)

        azimuth475 = step_results[step_results['cwl'] == 475]['azimuth'].values[0]
        azimuth550 = (azimuth475 + 45)
        if azimuth550 >= 360:
            azimuth550 -= 360
        norm550 = step_results[step_results['azimuth'] == azimuth550]['reflectance'].values[0]

        if normalise:
            step_results['reflectance'] = step_results['reflectance'] / norm550
            step_results['stddev'] = step_results['stddev'] / norm550
            y_label = '550 nm Normalised Reflectance'
            title = 'Normalised Reflectance vs Wavelength for each Step'
            filename = 'plot_reflectance_vs_cwl_per_step_normalised.png'
        else:
            y_label = 'Reflectance'
            title = 'Reflectance vs Wavelength for each Step'
            filename = 'plot_reflectance_vs_cwl_per_step.png'

        # drop the other 550 nm
        drop550 = step_results[(step_results['cwl'] == 550) & (step_results['azimuth'] != azimuth550)]
        step_results.drop(drop550.index, inplace=True)

        ax.errorbar(
            step_results['cwl'], 
            step_results['reflectance'], 
            yerr=step_results['stddev'], 
            fmt='o-',
            capsize=5,
            color=color[c],
            label=step.split('_')[1])
        
        ax.errorbar(
            drop550['cwl'], 
            drop550['reflectance'], 
            yerr=drop550['stddev'], 
            fmt='x-',
            capsize=5,
            color=color[c])
        
        ax.set_xlabel('Wavelength (nm)')
        ax.set_ylabel(y_label)
        ax.set_title(title)

    for i in np.arange(0, len(results)):
        lo_cwl = results['cwl'][i] - results['fwhm'][i]/2
        hi_cwl = results['cwl'][i] + results['fwhm'][i]/2
        ax.axvspan(lo_cwl, hi_cwl, alpha=0.2, color=cam_cols[results['Device'][i]]) 

    handles, labels = plt.gca().get_legend_handles_labels()
    point = Line2D([0], [0], label='Spare', marker='x', color='k', linestyle='None')
    handles.extend([point])
    sns.despine(ax=ax)

    ax.legend(handles=handles, loc='center left', bbox_to_anchor=(1, 0.5), title='Step', fontsize='x-small')
    fig.tight_layout()
    fig.show()

    fig.savefig(Path(filename).resolve(), dpi=300)

    return fig, ax
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np

def plot_reflectance_vs_cwl_per_phase(concat_results, normalise):

    fig, ax = plt.subplots(figsize=(5,3.4), dpi=300)

    phases = sorted(concat_results['phase'].unique())
    azimuths = sorted(concat_results['azimuth'].unique())

    phase_dict = dict(zip(phases, range(len(phases))))

    color = plt.cm.coolwarm(np.linspace(0, 1, len(phases)))

    azi_marks = ['o', 'x', '+', 'D']

    for phase in phases:
        step_results = concat_results[concat_results['phase'] == phase].copy() 
        step_results.sort_values(by='cwl', inplace=True)

        azimuths = sorted(step_results['azimuth'].unique())

        for m, azimuth in enumerate(azimuths):
            azimuth_results = step_results[step_results['azimuth'] == azimuth].copy()
            azimuth_results.sort_values(by='cwl', inplace=True)     

            norm550 = azimuth_results[azimuth_results['cwl'] == 550]['reflectance'].mean()

            if normalise:
                azimuth_results['reflectance'] = azimuth_results['reflectance'] / norm550
                azimuth_results['stderr'] = azimuth_results['stderr'] / norm550
                azimuth_results['stddev'] = azimuth_results['stddev'] / norm550
                y_label = 'Normalised Reflectance Factor'
                title = 'Normalised Reflectance vs Wavelength for each Phase'
                filename = 'plot_reflectance_vs_cwl_per_phase_normalised.png'
            else:
                y_label = 'Reflectance Factor'
                title = 'Reflectance vs Wavelength for each Phase'
                filename = 'plot_reflectance_vs_cwl_per_phase.png'
            
            ax.errorbar(
                azimuth_results['cwl'], 
                azimuth_results['reflectance'], 
                yerr=azimuth_results['stddev'], 
                marker=azi_marks[m],
                capsize=2,
                color='k')
            
            ax.errorbar(
                azimuth_results['cwl'], 
                azimuth_results['reflectance'], 
                yerr=azimuth_results['stderr'], 
                marker=azi_marks[m],
                capsize=5,
                color=color[phase_dict[phase]],
                label=f"{phase} °, {azimuth} °")

        
        ax.set_xlabel('Wavelength (nm)')
        ax.set_ylabel(y_label)
        ax.set_title(title)

    for i in np.arange(0, len(results)):
        lo_cwl = results['cwl'][i] - results['fwhm'][i]/2
        hi_cwl = results['cwl'][i] + results['fwhm'][i]/2
        ax.axvspan(lo_cwl, hi_cwl, alpha=0.2, color=cam_cols[results['Device'][i]]) 

    handles, labels = plt.gca().get_legend_handles_labels()
    point = Line2D([0], [0], label='Spare', marker='x', color='k', linestyle='None')
    handles.extend([point])
    sns.despine(ax=ax)

    ax.legend(handles=handles, loc='center left', bbox_to_anchor=(1, 0.5), title=f'$g$, $\Psi$', fontsize='xx-small')
    fig.tight_layout()
    fig.show()

    fig.savefig(Path(filename).resolve(), dpi=300)

    return fig, ax
```

```python
bands = pos1_results['Device'].unique()

azi_marks = ['o', 'x', '+', 'D']

r_30_cwl = []

for c, band in enumerate(bands):
    band_results = filter_rotation_results[filter_rotation_results['devices'] == band].copy()
    cwl = band_results['cwl'].values[0]
    x = band_results['phase']
    y = band_results['reflectance']

    # linear fit
    m, b = np.polyfit(x, y, 1)

    r = pos1_results[pos1_results['cwl'] == cwl]['Mean of Single-Frame ROI'].values[0]

    # phase correction given reflectance and change in phase
    phase = pos1_results[pos1_results['cwl'] == cwl]['Phase'].values[0]
    r_30 = m*(30 - phase) + r

    r_30_cwl.append(r_30)

pos1_results['Phase Corrected Mean of Single-Frame'] = r_30_cwl
```
